## 4. 字符串与格式化输出

### 4.1 索引与切片

字符串是一串字符的序列，每个字符有编号（**索引**），**从0开始**。索引表示"距起点多远"，第1个字符距离为0。

````python
code = "600519"
print(code[0])      # '6'      第1个字符
print(code[-1])     # '9'      倒数第1个字符
print(code[0:3])    # '600'    切片：含头不含尾
print(code[3:])     # '519'
print(len(code))    # 6        长度
````

### 4.2 拼接与常用方法

**方法**是写在对象后面、用点号调用的"功能"。字符串是**不可变**的：方法返回新字符串，原字符串不变。

````python
code = "600519"
exchange = "SH" + code      # 拼接：'SH600519'
print(exchange)
print("ha" * 3)             # 重复：'hahaha'

name = "Kweichow Moutai"
print(name.lower())         # 转为小写
print(name.upper())         # 转为大写
print("2026-08-04".split("-"))          # 按分隔符切开
print(", ".join(["600519", "000858"]))  # 用分隔符拼接
````

### 4.3 f-string：最好用的格式化输出

- 字符串前加 `f`，`{变量}` 会被替换为变量的值；

````python
name = "贵州茅台"
price = 1500.35
ret = 0.0752

print(f"{name} 当前价格 {price} 元")
print(f"日涨幅为 {ret:.2%}")      # 百分比，保留2位小数
print(f"价格保留两位小数：{price:.2f}")
````

### 6.2C 列表推导式：从已有循环缩写

先理解循环，再使用推导式。读法是“对每个 score，若达到 80，就保留 score”。复杂流程仍分步写。

````python
scores = [82, 95, 68]
passed = []
for score in scores:
    if score >= 80:
        passed.append(score)
short_form = [score for score in scores if score >= 80]
print(passed, short_form)
````

### 6.3 例：批量计算日收益率

模式：**准备空列表 → 循环计算 → append 收集结果**，非常常用。（预告：学完 NumPy 后，这段代码可以一行搞定。）

````python
prices = [100.0, 102.0, 101.0, 105.0]
returns = []                    # 空列表，用于存放结果

for i in range(1, len(prices)):
    r = (prices[i] - prices[i-1]) / prices[i-1]
    returns.append(r)           # 把结果追加进列表

print(returns)    # [0.02, -0.0098..., 0.0396...]
````

### 6.4 while 循环与 break / continue

`while` 循环必须保证条件终会变为 `False`，否则程序会"死循环"卡住。

````python
# while：条件为 True 就一直执行
value = 100
years = 0
while value < 200:        # 复利翻倍需要几年？
    value = value * 1.05
    years = years + 1
print(years)              # 15
````

````python
for i in range(10):
    if i == 3:
        continue          # 跳过本次，进入下一次
    if i == 6:
        break             # 立即结束整个循环
    print(i)
````

## 7. 函数：把常用计算"打包"

函数的三要素：**名字**（做什么） + **参数**（输入） + **返回值**（输出）。

### 7.1 def 定义函数

三引号字符串 `"""..."""` 是**文档字符串**：说明函数用途。没有 `return` 的函数返回 `None`。

````python
def simple_return(p_prev, p_now):
    """计算简单收益率"""
    return (p_now - p_prev) / p_prev

r = simple_return(100, 105)
print(r)              # 0.05
print(f"{r:.2%}")     # 5.00%
````

### 7.2 默认参数与关键字参数

金融公式（现值、终值、年金）特别适合封装成函数。

````python
def future_value(pv, rate, years=1):
    """复利终值：years 默认为 1"""
    return pv * (1 + rate) ** years

print(future_value(100, 0.05))              # 105.0
print(future_value(100, 0.05, 10))          # 162.89
print(future_value(100, rate=0.05, years=3))  # 指明参数名
````

### 7.4 print 与 return 的区别

print 负责展示，return 把结果交给调用者。需要继续计算时，应当返回结果。没有 return 的函数调用结果是 None。

````python
def show_double(number):
    print(number * 2)

def double(number):
    return number * 2

shown = show_double(3)
returned = double(3)
print(shown, returned + 1)  # None 7
````

### 8.1 问题：怎样管理多个矩形？

已知宽和高，怎样求面积？先沿用第2章的字典和函数。长度单位为厘米，面积单位为平方厘米。

````python
def rectangle_area(rect):
    return rect["width"] * rect["height"]

rect_a = {"width": 3, "height": 4}
rect_b = {"width": 5, "height": 2}
print(rectangle_area(rect_a), rectangle_area(rect_b))
````

**输出解释**：手算得到 3 × 4 = 12、5 × 2 = 10。字典保存数据，函数接收数据并计算；当相关操作增多时，可以用类组织这些数据和操作。

### 8.2 类、对象与实例：先分清三个词

类（class）定义一类对象的属性与行为，例如 Rectangle。对象（object）是程序中具体存在的值；实例强调“这个对象属于哪个类”。

一个具体矩形是 Rectangle 的实例。宽、高是属性（数据），求面积是方法（操作）。同一个类可以创建多个对象，每个对象保存自己的宽和高。

学习目标：读懂“对象.属性”和“对象.方法()”，写出简单的类，再用股票持仓练习。前面学过的变量、函数、字典和循环，正好是本节的基础。

### 8.3 第一个类：矩形的宽、高与面积

class 后面是类名；类体中的 def 定义方法。按缩进看结构：两个方法并列，方法内部再缩进。

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
````

**输出解释**：这里只定义规则，没有创建具体矩形，因此没有输出。__init__ 初始化实例属性；area 读取这些属性并返回面积。

### 8.4 创建两个实例：同一套规则，不同数据

调用 Rectangle(宽, 高) 创建实例。变量 a、b 分别引用两个不同的矩形对象。

````python
a = Rectangle(3, 4)
b = Rectangle(5, 2)
print(a.width, a.height, a.area())
print(b.width, b.height, b.area())
print(isinstance(a, Rectangle))
````

**输出解释**：第一行是 a 的宽、高、面积；第二行是 b 的。isinstance 检查对象是否属于给定类或其子类。类定义通常只写一次，实例可以创建很多个。

### 8.5 self：这一次操作的是哪个对象？

实例方法的第一个参数通常叫 self，表示当前实例。通过 a.area() 调用时，Python 自动把 a 传入该参数。

````python
print(a.area())
print(Rectangle.area(a))
print(b.area())
````

**输出解释**：前两行在这里等价，都是读取 a 的宽和高；第三行读取 b 的。self 是约定的参数名，不是关键字。调用 a.area() 时不用再手动传 a。

### 8.6 __init__：把输入保存到实例中

以 a = Rectangle(3, 4) 为例：实例创建后，__init__ 接收到 self、width=3、height=4。

self.width = width：右侧 width 是本次调用的参数；左侧 self.width 是保存在实例上的属性。初始化结束后，仍能通过 a.width 访问它。

__init__ 是初始化方法，不负责返回实例；不要在其中返回业务结果。默认对象创建机制还涉及 __new__，本节不展开。

### 8.7 属性与方法：什么时候加括号？

读取属性用点号；执行方法用点号和圆括号。方法本身也是一个可被引用的对象。

````python
print(a.width)
print(a.area())
area_function = a.area
print(callable(area_function))
print(area_function())
````

**输出解释**：a.area 取得绑定到 a 的方法，a.area() 才执行它。a.width 保存整数，写成 a.width() 会因整数不可调用而报 TypeError。

### 8.8 方法返回结果，还是修改状态？

area() 返回一个计算结果，不修改矩形。直接给属性赋值会改变对象的状态；下一次计算使用新状态。

````python
a.width = 6
result = a.area()
print(result)
print(b.area())
````

**输出解释**：a 的面积变成 6 × 4 = 24；b 仍是 10。return 使结果可以继续参与运算；只有 print 而没有 return 的方法默认返回 None。

### 8.10 赋值不会复制对象

````python
original = Rectangle(3, 4)
alias = original
independent = Rectangle(3, 4)
alias.width = 8
print(original.area(), independent.area())
print(alias is original)
print(independent is original)
````

### 8.11 问题：一笔持仓现在值多少钱？

现在把矩形的“数据＋操作”迁移到持仓。一个 StockPosition 实例表示一只股票的一笔持仓。全部数字均为自编教学数据，不需要下载行情。

symbol 是代码字符串；shares 是股数（股）；cost_price 是每股买入成本；price 是当前每股价格，价格单位均为元/股。

手算：100 股，成本 10 元，现价 12 元。市值为 100 × 12 = 1200 元，未实现盈亏为 100 × (12 − 10) = 200 元。暂不计费用、分红及交易。

### 8.12 持仓类：保存字段，定义两个计算方法

每个属性只保存基础信息；市值与盈亏在调用时由当前价格计算，避免另存一份容易过期的结果。

````python
class StockPosition:
    def __init__(self, symbol, shares, cost_price, price):
        self.symbol = symbol
        self.shares = shares
        self.cost_price = cost_price
        self.price = price

    def market_value(self):
        return self.shares * self.price

    def unrealized_pnl(self):
        return self.shares * (self.price - self.cost_price)

    def update_price(self, new_price):
        self.price = new_price
````

**输出解释**：market_value 和 unrealized_pnl 返回金额；update_price 修改状态，没有显式 return，因此返回 None。教学示例假定股数与价格为非负数。 这也是封装：将持仓数据和相关操作放在同一个类中，提供统一的调用接口；本例属性仍可直接访问。

### 8.13 创建持仓：输出每个数字的含义

### 8.15 对象列表：用同一种方法处理多笔持仓

列表的每个元素是一笔持仓对象，循环中 p 指向当前元素。本例重新创建数据，避免依赖前面改过的价格。

````python
positions = [
    StockPosition("000001", 100, 10.0, 12.0),
    StockPosition("600000", 200, 8.0, 7.5),
]
total_value = 0
total_pnl = 0
for p in positions:
    print(p.symbol, p.market_value(), p.unrealized_pnl())
    total_value += p.market_value()
    total_pnl += p.unrealized_pnl()
print(total_value, total_pnl)
````

### 8.16 继承：在已有类上增加一种能力（选学）

Square 是一种特殊的 Rectangle。通过继承复用 area，通过 super() 调用父类初始化，避免重复写面积公式。

````python
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)

square = Square(3)
print(square.area())
print(isinstance(square, Rectangle))
````

**输出解释**：边长 3，面积为 9。这个简化实现没有阻止分别修改宽和高，不能保证修改后仍为正方形；本例只演示继承与初始化。

### 8.17 重写方法：同名操作可以有不同实现（选学）

子类可以重新定义父类已有的方法。调用哪个实现，取决于当前对象的类。

````python
class Greeter:
    def greet(self):
        return "Hello"

class ChineseGreeter(Greeter):
    def greet(self):
        return "你好"

for speaker in [Greeter(), ChineseGreeter()]:
    print(speaker.greet())
````

**输出解释**：循环使用相同的 greet() 调用，得到不同问候语。这是多态的一个例子：同一个接口由不同对象实现。

### 8.18 多态也可以不依赖共同父类（选学）

Python 常按对象能否提供所需方法来使用它。下面两个类没有自定义的共同父类，但都有 market_value()。

````python
class Cash:
    def __init__(self, amount):
        self.amount = amount

    def market_value(self):
        return self.amount

assets = [StockPosition("000001", 100, 10.0, 12.0),
          Cash(800.0)]
print(sum(asset.market_value() for asset in assets))
````

**输出解释**：合计 2000 元。调用者关心的是能否按同一含义返回以元计的市值；这种使用方式通常称为鸭子类型，方法名相同还必须有一致的业务含义。

### 8.19 组合：一个组合持有多项资产（选学）

Portfolio 不是一种 StockPosition，而是包含多个资产对象。用组合关系组织“有一个/有多个”。

````python
class Portfolio:
    def __init__(self, assets):
        self.assets = list(assets)

    def market_value(self):
        return sum(a.market_value() for a in self.assets)

portfolio = Portfolio(assets)
print(portfolio.market_value())
assets[0].update_price(11.0)
print(portfolio.market_value())
````

**输出解释**：市值从 2000 变成 1900 元。list(assets) 复制了外层列表，但内部仍是相同资产对象，所以更新股票价格会影响组合估值。

字典：少量字段记录也很直观，不必为了面向对象而写类。后续 NumPy、pandas 已提供成熟对象；批量表格运算仍优先学习数组和表格操作。

## 9. 综合案例：迷你股票分析

本章学的**变量、列表、循环、判断、f-string**在这里全部用上了。

````python
# 某股票近5日收盘价
prices = [100.0, 102.0, 101.0, 105.0, 103.0]

# 1. 区间收益率
total_ret = (prices[-1] - prices[0]) / prices[0]

# 2. 每日收益率，并统计上涨天数
up_days = 0
for i in range(1, len(prices)):
    r = (prices[i] - prices[i-1]) / prices[i-1]
    if r > 0:
        up_days += 1

print(f"区间收益率：{total_ret:.2%}")
print(f"上涨天数：{up_days} / {len(prices)-1}")
````
