## 4. 字符串与格式化输出

### 4.1 索引与切片

字符串是一串字符的序列，每个字符有编号（**索引**），**从0开始**。索引表示"距起点多远"，第1个字符距离为0。

````python
code = "600519"
print(code[0])      # '6'      第1个字符
print(code[-1])     # '9'      倒数第1个字符
print(code[0:3])    # '600'    切片：含头不含尾
print(code[3:])     # '519'
print(len(code))    # 6        长度
````

### 4.2 拼接与常用方法

**方法**是写在对象后面、用点号调用的"功能"。字符串是**不可变**的：方法返回新字符串，原字符串不变。

````python
code = "600519"
exchange = "SH" + code      # 拼接：'SH600519'
print(exchange)
print("ha" * 3)             # 重复：'hahaha'

name = "Kweichow Moutai"
print(name.lower())         # 转为小写
print(name.upper())         # 转为大写
print("2026-08-04".split("-"))          # 按分隔符切开
print(", ".join(["600519", "000858"]))  # 用分隔符拼接
````

### 4.3 f-string：最好用的格式化输出

- 字符串前加 `f`，`{变量}` 会被替换为变量的值；
- 冒号后是格式说明：`.2f` 保留两位小数，`.2%` 百分比。

````python
name = "贵州茅台"
price = 1500.35
ret = 0.0752

print(f"{name} 当前价格 {price} 元")
print(f"日涨幅为 {ret:.2%}")      # 百分比，保留2位小数
print(f"价格保留两位小数：{price:.2f}")
````

## 5. 基本数据结构

一个现实问题：如果一只股票有 250 个交易日收盘价，难道要定义 250 个变量？我们需要能"一次装下一串数据"的容器。

| 结构 | 写法 | 可变？ | 典型用途 |
| --- | --- | --- | --- |
| 列表 list | `[1, 2, 3]` | 可变 | 一串股价、日期 |
| 元组 tuple | `(1, 2, 3)` | 不可变 | 固定的一行记录 |
| 字典 dict | `{"a": 1}` | 可变 | 代码→价格 的映射 |
| 集合 set | `{1, 2, 3}` | 可变 | 去重、成员判断 |

### 5.1 列表（list）：最常用的容器

- 方括号 `[]` 定义，元素用逗号分隔；
- 索引、切片规则与字符串一致（含头不含尾）；
- 列表**可以修改**。

````python
prices = [30.1, 30.5, 29.8, 31.2, 30.9]
codes  = ["600519", "000858", "601318"]
mixed  = [1, "two", 3.0, True]    # 可混装不同类型

print(prices[0])       # 30.1   索引从0开始
print(prices[-1])      # 30.9   最后一个
print(prices[1:3])     # [30.5, 29.8]  切片
print(len(prices))     # 5

prices[0] = 30.0       # 列表可以修改
prices
````

### 5.2 列表的增删与运算

注意：`append` 在原列表上修改；而 `prices + [31.0]` 生成**新**列表，原列表不变。

````python
codes = ["600519", "000858"]
codes.append("601318")      # 末尾追加
codes.insert(0, "000001")   # 在指定位置插入
codes.remove("000858")      # 按值删除
popped = codes.pop()        # 弹出最后一个
print(codes, popped)
````

````python
prices = [30.1, 30.5, 29.8]
print(prices + [31.0])     # 拼接成新列表
print([0] * 5)             # [0, 0, 0, 0, 0]
print("600519" in prices)  # False，成员判断

prices = [30.1, 30.5, 29.8]
prices.sort()              # 排序（升序）
prices
````

### 5.3 元组（tuple）：不可变的列表

圆括号 `()` 定义；一旦创建不能增删改，适合存放"不应被改动"的记录。

````python
record = ("600519", "贵州茅台", 1500.35)
print(record[0])          # '600519'

# record[1] = "茅台"      # 报错！元组元素不可修改，取消注释试试

# 元组解包：一次赋值给多个变量
code, name, price = record
print(code, name, price)
````

### 5.4 字典（dict）：按键取值

凡是"名称→数值"的对应关系（代码→收盘价、行业→市盈率），优先想到字典。

````python
prices = {"600519": 1500.35, "000858": 128.6, "601318": 55.2}
print(prices["600519"])          # 1500.35，按"键"取值
prices["000001"] = 12.5          # 新增或修改键值对
print("601318" in prices)        # True

print(list(prices.keys()))       # 所有键
print(list(prices.values()))     # 所有值
print(list(prices.items()))      # 所有(键, 值)对
````

### 5.5 集合（set）：自动去重

集合**无序、不重复**。常用于数据去重、快速判断成员。注意：`{}` 是空字典，空集合要用 `set()`。

````python
s = {"600519", "000858", "600519"}   # 自动去重
print(len(s))            # 2

s.add("601318")
print("600519" in s)     # True，成员判断非常快

a = {"600519", "000858"}
b = {"000858", "601318"}
print(a & b)             # 交集 {'000858'}
print(a | b)             # 并集
````

## 6. 控制流：让程序会"判断"与"重复"

### 6.1 if 条件判断

- **缩进是Python的语法**：缩进（4个空格）内的行属于该分支；
- 条件行末尾必须有**冒号** `:`；
- 缩进错误（漏缩进、空格与Tab混用）是最常见的报错来源。

````python
ret = 0.075

if ret > 0:
    print("今日上涨")
elif ret == 0:
    print("今日平盘")
else:
    print("今日下跌")
````

### 6.2 for 循环：遍历序列

`for 变量 in 序列:` 让循环变量依次等于序列中的每个元素；`range()` 常用来按次数循环，同样**含头不含尾**。

````python
codes = ["600519", "000858", "601318"]
for code in codes:          # 依次取出每个元素
    print(code)
````

````python
for i in range(5):          # range(5) 生成 0,1,2,3,4
    print(i)
````

````python
for i in range(1, 11):      # 1 到 10（含头不含尾）
    print(i)
````

### 6.3 例：批量计算日收益率

模式：**准备空列表 → 循环计算 → append 收集结果**，非常常用。（预告：学完 NumPy 后，这段代码可以一行搞定。）

````python
prices = [100.0, 102.0, 101.0, 105.0]
returns = []                    # 空列表，用于存放结果

for i in range(1, len(prices)):
    r = (prices[i] - prices[i-1]) / prices[i-1]
    returns.append(r)           # 把结果追加进列表

print(returns)    # [0.02, -0.0098..., 0.0396...]
````

### 6.4 while 循环与 break / continue

`while` 循环必须保证条件终会变为 `False`，否则程序会"死循环"卡住。

````python
# while：条件为 True 就一直执行
value = 100
years = 0
while value < 200:        # 复利翻倍需要几年？
    value = value * 1.05
    years = years + 1
print(years)              # 15
````

````python
for i in range(10):
    if i == 3:
        continue          # 跳过本次，进入下一次
    if i == 6:
        break             # 立即结束整个循环
    print(i)
````

## 7. 函数：把常用计算"打包"

函数的三要素：**名字**（做什么） + **参数**（输入） + **返回值**（输出）。

### 7.1 def 定义函数

三引号字符串 `"""..."""` 是**文档字符串**：说明函数用途。没有 `return` 的函数返回 `None`。

````python
def simple_return(p_prev, p_now):
    """计算简单收益率"""
    return (p_now - p_prev) / p_prev

r = simple_return(100, 105)
print(r)              # 0.05
print(f"{r:.2%}")     # 5.00%
````

### 7.2 默认参数与关键字参数

金融公式（现值、终值、年金）特别适合封装成函数。

````python
def future_value(pv, rate, years=1):
    """复利终值：years 默认为 1"""
    return pv * (1 + rate) ** years

print(future_value(100, 0.05))              # 105.0
print(future_value(100, 0.05, 10))          # 162.89
print(future_value(100, rate=0.05, years=3))  # 指明参数名
````

## 8. 综合案例：迷你股票分析

本章学的**变量、列表、循环、判断、f-string**在这里全部用上了。

````python
# 某股票近5日收盘价
prices = [100.0, 102.0, 101.0, 105.0, 103.0]

# 1. 区间收益率
total_ret = (prices[-1] - prices[0]) / prices[0]

# 2. 每日收益率，并统计上涨天数
up_days = 0
for i in range(1, len(prices)):
    r = (prices[i] - prices[i-1]) / prices[i-1]
    if r > 0:
        up_days += 1

print(f"区间收益率：{total_ret:.2%}")
print(f"上涨天数：{up_days} / {len(prices)-1}")
````
