## 练习3｜逐只计算与累加

打印每只股票市值，并计算总市值。给“计算”和“累加”两行加注释。

````python
# total = 0
# for item in holdings:
#     value = ...
#     total = ...
# print(total)
````

**我的解释：**（填写预测、实际结果或原因。）

<details><summary>参考答案（先自己动手）</summary>

```python
total = 0
for item in holdings:
    value = item["price"] * item["shares"]  # 计算本只股票市值
    print(item["name"], value)
    total = total + value  # 累加到总市值
print("总市值：", total)
```

三只市值分别为2000、2000、1500元，总计5500元。total必须在循环之前初始化。

</details>

````python
# for item in holdings:
#     value = ...
#     if ...:
#         print(item["name"], value)
````
