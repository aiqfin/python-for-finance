````python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 教师提供：从Notebook目录、上机练习目录或课程根目录运行均可。
cwd = Path.cwd()
if (cwd / "上机练习").is_dir():
    LAB = cwd / "上机练习"
elif cwd.name == "notebooks":
    LAB = cwd.parent
else:
    LAB = cwd
DATA = LAB / "data"
OUTPUTS = LAB / "outputs"
OUTPUTS.mkdir(exist_ok=True)

df = pd.read_csv(DATA / "stocks_2025.csv", dtype={"Stkcd": str}, parse_dates=["Trddt"])
plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "PingFang SC", "SimHei", "DejaVu Sans"]
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["figure.figsize"] = (9, 4)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.25
TARGET = "000001"
stock = df.loc[df["Stkcd"] == TARGET].sort_values("Trddt").set_index("Trddt")
````

````python
small_returns = pd.Series([0.10, -0.10])
print("直接相加：", small_returns.sum())
print("复合累计：", (1 + small_returns).prod() - 1)
print(stock["Dretwd"].head())
````

**输出解释：**先涨10%再跌10%，100元变为99元，累计-1%而不是0%。Dretwd是小数，不要在复利计算前乘100。

## 练习5｜累计收益与简单相加

计算累计收益曲线与期末累计收益，对比直接相加。解释cumprod与prod的差别。

````python
# r = stock["Dretwd"]
# wealth = (1 + r).cumprod()
# cumulative = ...
# 请绘图并比较期末值与r.sum()
````

**我的解释：**（填写预测、实际结果或原因。）

<details><summary>参考答案（先自己动手）</summary>

```python
r = stock["Dretwd"]
wealth = (1 + r).cumprod()
cumulative = wealth - 1
print(f"期末累计收益：{cumulative.iloc[-1]:.2%}；简单相加：{r.sum():.2%}")
fig, ax = plt.subplots()
ax.plot(stock.index, cumulative * 100, label=TARGET)
ax.set(title="2025年含红利累计收益｜含首日收益", xlabel="交易日期", ylabel="累计收益（%）")
ax.legend()
fig.tight_layout()
plt.show()
```

cumprod保留每一步连乘，prod只给总乘积。基准本金1位于首个样本日收益发生之前，因此图中第一个点不强制为0。不扣交易成本。

</details>
