## 准备与小示例

每行是一只示例股票一个观测日的行情，共15行。date为日期，code为代码，close为价格（元），volume为成交量（股）。全部为合成数据。

````python
import pandas as pd

df = pd.DataFrame({
    "date": list(pd.date_range("2025-01-06", periods=5, freq="B")) * 3,
    "code": ["A"] * 5 + ["B"] * 5 + ["C"] * 5,
    "close": [10, 11, 10.5, 12, 11.5, 20, 19, 20, 21, 20.5, 5, 5.1, 5, 5.2, 5.3],
    "volume": [1000, 1200, 900, 1500, 1100, 800, 900, 1000, 1200, 950, 2000, 2100, 1800, 2200, 2300],
})
````

````python
scores = pd.Series([80, 90], index=["小王", "小李"])
print(scores.loc["小李"])
display(df.head())
````

**输出解释：**loc按标签“小李”取到90。行情表前五行是A的五日记录，不是五只股票。

## 练习4｜多个条件

筛选A且价格大于11元的记录。把阈值改为10.5，行数如何变化？提示：每个条件加括号，用&连接。

````python
# selected = df.loc[(...) & (...)]
# display(selected)
````

**我的解释：**（填写预测、实际结果或原因。）

<details><summary>参考答案（先自己动手）</summary>

```python
selected = df.loc[(df["code"] == "A") & (df["close"] > 11)]
display(selected)
print(len(selected))
print(len(df.loc[(df["code"] == "A") & (df["close"] > 10.5)]))
```

阈值11时为2行，阈值10.5时为3行。>不包含等于边界的记录。

</details>

````python
base = pd.Series([10, 20], index=["A", "B"])
bonus = pd.Series([1, 2], index=["B", "A"])
# print(base + bonus)
````

**我的解释：**（填写预测、实际结果或原因。）

<details><summary>参考答案（先自己动手）</summary>

