## 准备与小示例

先比较列表与数组的乘法。价格矩阵每行是一个观测日，每列是一只股票，单位为元。

````python
import numpy as np

prices = np.array([[10.0, 20.0], [11.0, 19.0], [12.0, 20.0]])
````

````python
print([1, 2, 3] * 2)
print(np.array([1, 2, 3]) * 2)
````

**输出解释：**列表乘2重复元素；NumPy数组乘2使每个数字翻倍。
