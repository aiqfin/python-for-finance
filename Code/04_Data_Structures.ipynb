### Sets

```python
s = set(['u', 'd', 'ud', 'du', 'd', 'du'])
s
```

```python
t = set(['d', 'dd', 'uu', 'u'])
```

```python
s.union(t)  # all of s and t
```

```python
s.intersection(t)  # both in s and t
```

```python
s.difference(t)  # in s but not t
```

```python
t.difference(s)  # in t but not s
```

```python
s.symmetric_difference(t)  # in either one but not both
```
